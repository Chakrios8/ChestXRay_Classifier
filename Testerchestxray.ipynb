```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from google.colab import drive, files

# 1. Mount Drive to grab your pre-trained model
drive.mount('/content/drive')

# 2. Load your saved model in 2 seconds (NO retraining needed!)
best_model = keras.models.load_model('/content/drive/MyDrive/best_medical_cnn.keras')
print("✓ Pre-trained Medical CNN loaded successfully!")

# 3. Define the single image prediction function
def predict_single_xray(image_path, model, threshold=0.5):
    img = keras.utils.load_img(image_path, target_size=(224, 224))
    img_array = keras.utils.img_to_array(img)
    input_tensor = np.expand_dims(img_array, axis=0)

    logit = model.predict(input_tensor, verbose=0)[0][0]
    prob = tf.sigmoid(logit).numpy()
    diagnosis = "PNEUMONIA" if prob >= threshold else "NORMAL"
    confidence = prob if diagnosis == "PNEUMONIA" else (1.0 - prob)

    # Display image & diagnosis
    fig, axes = plt.subplots(1, 2, figsize=(10, 4))
    axes[0].imshow(img)
    axes[0].axis('off')
    axes[0].set_title("Input Chest Radiograph", fontsize=12)

    axes[1].barh(["Normal", "Pneumonia"], [1.0 - prob, prob],
                 color=['#16A34A', '#DC2626'], edgecolor='black', alpha=0.8)
    axes[1].set_xlim([0, 1])
    axes[1].set_xlabel("Probability")
    axes[1].axvline(threshold, color='black', linestyle='--', label=f'Threshold ({threshold})')
    axes[1].legend()
    axes[1].set_title("Clinical Confidence", fontsize=12)

    color = "red" if diagnosis == "PNEUMONIA" else "green"
    plt.suptitle(f"Diagnosis: {diagnosis} ({confidence*100:.1f}% Confidence)",
                 fontsize=14, fontweight='bold', color=color)
    plt.tight_layout()
    plt.show()

# 4. Upload and predict!
print("\n📤 Upload any Chest X-Ray to test:")
uploaded = files.upload()
img_file = list(uploaded.keys())[0]
predict_single_xray(img_file, best_model)
```
